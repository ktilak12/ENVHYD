```python

import matplotlib.pyplot as plt
import seaborn as sns


sns.set_style("whitegrid")
```

```python
df = pd.read_csv("rainfallEH.csv")
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.hist(figsize=(15,10))
plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(df.corr(numeric_only=True),
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))

# Melt the DataFrame to transform seasonal columns into 'Season' and 'Rainfall' columns
seasonal_cols = ['Jan-Feb', 'Mar-May', 'Jun-Sep', 'Oct-Dec']
df_melted = df.melt(value_vars=seasonal_cols, var_name='Season', value_name='Rainfall')

sns.barplot(x='Season',
            y='Rainfall',
            data=df_melted)

plt.show()
```
